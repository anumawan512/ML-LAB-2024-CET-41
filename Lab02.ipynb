```python
units = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
pop2017 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
p23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def mkDict(turns):
    return {str(karIdx): (x.item() if isinstance(x, np.generic) else x) for karIdx, x in turns}

allPop = p23.sum()
print(f"6a: total population 2023 = {allPop:.2f} million")

share = p23 / allPop * 100
print(f"6b: share of total (%): {mkDict(zip(units, np.round(share, 2)))}")

print(f"6c: most populous unit: {units[np.argmax(p23)]}")
print(f"6d: units above 20 million: {units[np.where(p23 > 20)]}")

growthRate = ((p23 / pop2017) ** (1 / 6) - 1) * 100
print(f"6e: annual growth 2017-2023 (% per year): {mkDict(zip(units, np.round(growthRate, 2)))}")
print(f"6f: fastest-growing unit: {units[np.argmax(growthRate)]} ({growthRate.max():.2f}% per year)")

national = ((p23.sum() / pop2017.sum()) ** (1 / 6) - 1) * 100
yearsToDouble = 70 / growthRate.max()
print(f"6 interpret: national growth = {national:.2f}% per year.")
print(f"6 interpret: {units[np.argmax(growthRate)]} would double in about {yearsToDouble:.0f} years at its current rate.")
print(f"6 interpret: AJK grew {growthRate[-1]:.2f}% per year vs national {national:.2f}% -> "
      f"AJK is {national - growthRate[-1]:.2f} points below the national rate (AJK is essentially flat in this sample).")
```

```python
towns = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
months = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temps = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
rainData = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def mkDict(turns):
    return {str(karIdx): (x.item() if isinstance(x, np.generic) else x) for karIdx, x in turns}

print(f"7a: shape of temp (cities, months): {temps.shape}")

yearAvg = temps.mean(axis=1)
print(f"7b: yearly mean temperature (deg C): {mkDict(zip(towns, np.round(yearAvg, 1)))}")

warmest = months[temps.argmax(axis=1)]
print(f"7c: hottest month per city: {mkDict(zip(towns, warmest))}")

print(f"7d: 5-city monthly average (deg C): {mkDict(zip(months, np.round(temps.mean(axis=0), 1)))}")

above35 = np.sum(temps > 35, axis=1)
print(f"7e: months above 35 deg C: {mkDict(zip(towns, above35))}")

lahoreFahr = temps[0] * 1.8 + 32
print(f"7f: Lahore monthly temperature (deg F): {np.round(lahoreFahr, 1)}")

mons = rainData[:, 6:9].sum(axis=1)
print(f"7g: monsoon rain Jul-Sep (mm): {mkDict(zip(towns, mons))}")
print(f"7g: wettest monsoon city: {towns[np.argmax(mons)]} ({mons.max()} mm)")

share = 100 * mons / rainData.sum(axis=1)
print(f"7h: monsoon share of annual rain (%): {mkDict(zip(towns, np.round(share, 1)))}")

karIdx = list(towns).index("Karachi")
print(f"7 interpret: Karachi has the highest yearly mean ({yearAvg.max():.1f} deg C) "
      f"but {above35[karIdx]} months above 35 deg C, while cooler-on-average Peshawar has {above35[4]} "
      f"and Lahore {above35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

```python
fy = np.array(["FY2023", "FY2024", "FY2025"])
fiscalMonths = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

cpi = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def mkDict(turns):
    return {str(karIdx): (x.item() if isinstance(x, np.generic) else x) for karIdx, x in turns}

yrAvg = cpi.mean(axis=1)
print(f"8a: average inflation per year (%): {mkDict(zip(fy, np.round(yrAvg, 2)))}")

fy24Row = cpi[1]
print(f"8b: FY2024 months above 25 %: {fiscalMonths[fy24Row > 25]} -> {np.sum(fy24Row > 25)} months")

change = cpi[:, 1:] - cpi[:, :-1]
print(f"8c: month-on-month change (percentage points; negative = drop):\n{np.round(change, 1)}")

biggestDrop = np.argmin(change, axis=1)
print(f"8d: month of biggest drop: {mkDict(zip(fy, fiscalMonths[biggestDrop + 1]))} | size (pp): {np.round(change.min(axis=1), 1)}")

fy25Vals = cpi[2]
tags = np.where(fy25Vals < 5, "Low", np.where(fy25Vals < 10, "Moderate", "High"))
print(f"8e: FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fiscalMonths, fy25Vals, tags))}")

print(f"8 interpret: average inflation went {yrAvg[0]:.1f}% -> {yrAvg[1]:.1f}% -> {yrAvg[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

```python
studentMarks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
rollno = np.arange(1, 21)
PASS_MARK = 40

graced = np.clip(studentMarks + 5, 0, 100)
print(f"9a: original marks: {studentMarks}")
print(f"9a: marks after +5 grace (cap 100): {graced}")

pBefore = np.mean(studentMarks >= PASS_MARK) * 100
afterPct = np.mean(graced >= PASS_MARK) * 100
print(f"9b: pass % before grace = {pBefore:.1f} %, after grace = {afterPct:.1f} %")

onlyGrace = np.logical_and(studentMarks < PASS_MARK, graced >= PASS_MARK)
print(f"9c: passed only because of grace (roll numbers): {rollno[onlyGrace]}")
print(f"9c: their original marks: {studentMarks[onlyGrace]}")

zscores = (graced - graced.mean()) / graced.std()
print(f"9d: mean = {graced.mean():.2f}, std = {graced.std():.2f}")
print(f"9d: z-scores (std devs from mean): {np.round(zscores, 2)}")
```

```python
pixels = np.zeros((8, 8), dtype=np.uint8)
pixels[1, 1:7] = 255
for r, c in zip([2, 3, 4, 5, 6], [6, 5, 5, 4, 4]):
    pixels[r, c] = 255

view01 = lambda a: (a > 0).astype(int)

print(f"10a: the '7' (1 = white):\n{view01(pixels)}")

flipImg = pixels[:, ::-1]
print(f"10b: flipped left-right:\n{view01(flipImg)}")

inverted = 255 - pixels
print(f"10c: inverted (white <-> black):\n{view01(inverted)}")

crop = pixels[:7, 1:]
print(f"10d: cropped 7x7 region, shape: {crop.shape}\n{view01(crop)}")

imgBright = np.clip(pixels.astype(np.int16) + 60, 0, 255).astype(np.uint8)
print(f"10e: brightened (+60, capped at 255); unique values: {np.unique(imgBright)}")

print(f"10f: shape = {pixels.shape} | dtype = {pixels.dtype} | white pixels = {np.sum(pixels == 255)}")
```
