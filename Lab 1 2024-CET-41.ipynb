```python
def find_dog(text):
    return "dog" in text.lower()

print(find_dog("Is there a dog here?"))
print(find_dog("DOG"))
print(find_dog("No pets here"))
```

```python
def count_dog(text):
    return text.lower().count("dog")

print(count_dog("This dog runs faster than that other dog"))
print(count_dog("Dog DOG dog"))
```

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        limit_add = 5
    else:
        limit_add = 0

    if speed <= 60 + limit_add:
        return "No Ticket"
    elif speed <= 80 + limit_add:
        return "Small Ticket"
    else:
        return "Big Ticket"

print(caught_speeding(60, False))
print(caught_speeding(65, True))
print(caught_speeding(65, False))
print(caught_speeding(85, True))
print(caught_speeding(85, False))
```
